## Formula para calcular area
$A = b * h/2$
---
## Datos de entrada

```python
# calcular area de un triangulo
# variables b,h, A
# formulas A = b * h/2
# Capturar el valor de A
print("Calcular area de un triangulo")
b = float(input("Ingrese el valor de la base: "))
h = float(input("Ingrese el valor de la altura: "))
A = b * h/2
print("El area del triangulo es: ", A)
print("Sebastian Farfan NL = 16")
```
